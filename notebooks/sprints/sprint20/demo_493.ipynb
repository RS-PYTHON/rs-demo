## Showcase of new filtering capabilities implemented in RPSY-493.  

cql2-text filter parameter was updated to handle multiple query parameters and operations. Previously we were able to search products (AUXIP/CADIP) using only publication dates / product names / session ID. Now queryables include: product:type, platform, constellation, Name, auxip:id, created, datetime, start_datetime, end_datetime, processing:datetime, processing:facility, published for ADGS and id, platform, constellation, sat:absolute_orbit, datetime, end_datetime, start_datetime, published, cadip:id, cadip:num_channels, cadip:station_unit_id, cadip:front_end_id, cadip:retransfer
cadip:antenna_status_ok, etc.

```python
# Fitering ADGS:
import requests
import pprint

BASE_ADGS = "http://rs-server-adgs:8000/auxip"
BASE_CADIP = "http://rs-server-cadip:8000/cadip"
pprint.pp(requests.get(f"{BASE_ADGS}/queryables").json())
```

```python
pprint.pp(requests.get(f"{BASE_ADGS}/search?filter=processing:facility=FOS").json())
```

query filters cand be even extendend using AND operator. Search AUX products from platform s1a with a specific date of publication.

```python
pprint.pp(requests.get(f'{BASE_ADGS}/search?filter=platform="sentinel-1a" AND datetime="2019-02-16T18:29:37.522Z"').json())
```

### Filters can now be applied to /items endpoints, with operator predefined collections, to enrich query capabilities.  
In the example below, we use predefined collection adgs_aux_pp2 that returns only products with type AUX_PP2, and we select also products comming from "FOS" processing center.

```python
pprint.pp(requests.get(f"{BASE_ADGS}/collections/adgs_aux_pp2/items/?filter=processing:facility='FOS'").json())
```

Request all products with type **AUX_PP2** sorted by eviction date. (Other ADGS sortables include: id (Name), auxip:id, file:size, type, created, start / end_datetime.


```python
pprint.pp(requests.get(f"{BASE_ADGS}/search?filter=product:type=AUX_PP2&sortby=-eviction_datetime").json())
```

CADIP sort parameters include: **id, start_datetime, datetime, end_datetime, published, platform, publicationDate, cadip:id, cadip:num_channels, cadip:station_unit_id, sat:absolute_orbit, cadip:acquisition_id, cadip:antenna_id, cadip:front_end_id, cadip:retransfer, cadip:antenna_status_ok, cadip:front_end_status_ok, cadip:planned_data_start, cadip:planned_data_stop, cadip:downlink_status_ok, cadip:delivery_push_ok**

```python
pprint.pp(requests.get(f'{BASE_CADIP}/search?filter=constellation="sentinel-2" AND cadip:retransfer=False&sorbtby=-published').json())
```

Request all **Sentinel 2** products from station **MTI**, with **retransfer property False**, ordered descending by **publicationdate**.

```python
pprint.pp(requests.get(f'{BASE_CADIP}/collections/mti_sentinel2/items?filter=cadip:retransfer=False&sorbtby=-published').json())
```
