```python
# Do some init
import os
import json
import requests

rs_server_href = os.environ["RSPY_WEBSITE"]
```

```python
# Call the endpoint that synchronises RS users registered in Keycloak and the linked users from the Cloud provider (OVH)
# This endpoint is also a trigger for gathering the keycloak user attributes, roles and bucket configuration
response = requests.post (f"{rs_server_href}/storage/accounts/update")
print(f"status_code = {response.status_code}")
print(response.content)
```

```python
# Call the endpoint to get the OBS access rights of user osam
response = requests.get (f"{rs_server_href}/storage/account/osam/rights")
print(f"status_code = {response.status_code}")
print(response.content)
```
