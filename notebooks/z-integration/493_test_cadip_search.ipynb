```python
import requests
import json

# Base URL for the API
BASE_URL = "http://rs-server-cadip:8000/cadip/collections/cadip_sentinel1/items"

# Mapping of properties to example values from the dataset
search_properties = {
    "cadip:id": "a5e9d3b8-7c4f-4a92-b76a-fa09e1e1b59c",
    "cadip:acquisition_id": "53186_1",
    "cadip:antenna_id": "MSP21",
    "cadip:antenna_status_ok": "true",
    "cadip:delivery_push_ok": "true",
    "cadip:downlink_status_ok": "true",
    "cadip:front_end_id": "01",
    "cadip:front_end_status_ok": "true",
    "cadip:num_channels": "2",
    "cadip:planned_data_start": "2020-01-05T07:22:04.051Z",
    "cadip:planned_data_stop": "2020-01-05T07:31:04.051Z",
    "cadip:retransfer": "false",
    "cadip:station_unit_id": "01",
    "constellation": "S1A",
    "datetime": "2020-01-05T18:52:26.165Z",
    "end_datetime": "2020-01-05T07:42:04.051Z",
    "id": "S1A_20200105072204051312",
    "platform": "S1A",
    "published": "2020-01-05T18:52:26.165Z",
    "sat:absolute_orbit": "53186",
    "start_datetime": "2020-01-05T07:22:04.051Z"
}

# Properties that contain date values
date_properties = {"datetime", "end_datetime", "published", "start_datetime", "cadip:planned_data_start", "cadip:planned_data_stop"}

# Counters for total requests and failures
total_requests = 0
total_failures = 0
failed_requests = []

# Generate all possible pairs of search properties
keys = list(search_properties.keys())
for i in range(len(keys)):
    for j in range(i + 1, len(keys)):
        prop1, prop2 = keys[i], keys[j]
        value1, value2 = search_properties[prop1], search_properties[prop2]
        
        # Wrap date values in double quotes
        formatted_value1 = f'"{value1}"' if prop1 in date_properties else value1
        formatted_value2 = f'"{value2}"' if prop2 in date_properties else value2
        
        url = f"{BASE_URL}?filter={prop1}={formatted_value1}&filter={prop2}={formatted_value2}"
        print(f"Requesting: {url}")
        total_requests += 1
        
        try:
            response = requests.get(url)
            response.raise_for_status()
            data = response.json()
            
            # Count how many items are in "features"
            num_items = len(data.get("features", []))
            if num_items > 0:
                print(f"Data found for {prop1} ({value1}) and {prop2} ({value2}): {num_items} items")
            else:
                print(f"No data found for {prop1} ({value1}) and {prop2} ({value2})")
        except requests.exceptions.RequestException as e:
            print(f"❌ Request failed for {prop1} and {prop2}: {e}")
            total_failures += 1
            failed_requests.append(str(e))
        print("-" * 80)
# Summary report
print("Summary:")
print(f"Total requests: {total_requests}")
print(f"Total failed requests: {total_failures}")
if failed_requests:
    print("Failed requests details:")
    for error in failed_requests:
        print(error)
```
