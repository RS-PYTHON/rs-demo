```python
import requests
import json

BASE_URL = "http://rs-server-adgs:8000/auxip/search"

search_properties = {
    "datetime": "2024-02-14T02:08:13.372Z",
    "created": "2024-02-14T02:08:13.372Z",
    "start_datetime": "2024-02-14T02:08:13.372Z",
    "end_datetime": "2024-02-14T02:08:23.372Z",
    "platform": "sentinel-1a",
    "constellation": "sentinel-1",
    "auxip:id": "9839aacf-828a-4157-837d-77b63c557ff1",
    "published": "2024-02-14T02:08:13.372Z",
    "product:type": "OPER_AUX_RESORB_OPOD",
    "processing:datetime": "2024-02-14T00:00:00.000Z",
    "processing:facility": "FOS"
}

for prop, value in search_properties.items():
    url = f"{BASE_URL}?filter={prop}={value}"
    print(f"Requesting: {url}")
    try:
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()
        
        # Check if features is not empty
        num_items = len(data.get("features", []))
        if num_items>0:
            print(f"Data found for {prop} ({value}): {num_items} items")
        else:
            print(f"❌ No data found for {prop} ({value})")
    except requests.exceptions.RequestException as e:
        print(f"❌ Request failed for {prop}: {e}")
    print("-" * 80)
```

```python
#
BASE_URL = "http://rs-server-adgs:8000/auxip/search"

urls = [f"{BASE_URL}?filter=processing:facility=PDMC&sortby=-processing:datetime&limit=5",
        f"{BASE_URL}?filter=processing:facility=FOS&sortby=-created"]
for url in urls:
    print(f"Requesting: {url}")
    try:
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        # Count how many items are in "features"
        num_items = len(data.get("features", []))
        if num_items > 0:
            print(f"Data found for {url}: {num_items} items")
        else:
            print(f"No data found for {url}")
    except requests.exceptions.RequestException as e:
        print(f"❌ Request failed for {url}: {e}")
    print("-" * 80)
```
